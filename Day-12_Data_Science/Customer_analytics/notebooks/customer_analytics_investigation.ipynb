## Data Quality Assessment

### Dataset Profile

- **Rows:** 10,020
- **Columns:** 9
- **Grain:** One row is intended to represent one customer.
- **Business Key:** `customer_id`
- **Unique customers:** 10,000

The dataset contains 10,020 records but only 10,000 unique customer IDs, indicating that duplicate customer records are present.

The dataset contains:
- 4 categorical attributes: `customer_id`, `region`, `customer_type`, `acquisition_channel`
- 5 numerical attributes: `tenure_months`, `orders`, `discount_pct`, `returns`, `revenue`

### Missing Values

- `discount_pct` contains **50 missing values**, representing approximately **0.50%** of the dataset.
- All other columns contain no missing values.
- The missing discount values should not be blindly imputed before understanding their impact on discount-based analysis.

### Duplicate Records

- **20 completely duplicated rows** were identified.
- **20 duplicate `customer_id` records** were identified.
- Since the intended grain is one row per customer, these duplicates violate the expected business grain.
- The duplicated records appear to be exact copies of existing customer records rather than separate customer activity.

Therefore, the duplicate records should be investigated and handled before calculating customer-level aggregates or business metrics.

### Business Rule Violations

- **Orders:** No invalid negative order values were identified.
- **Returns:** No negative returns or cases where returns exceeded orders were identified.
- **Discount:** No values outside the expected 0–100% range were identified.
- **Revenue:** No negative revenue values were identified.
- **Tenure:** No customers with zero or negative tenure were identified.

The core numerical fields therefore satisfy the basic business validity rules tested.

### Business Sanity Checks

A derived metric, `revenue_per_order`, was calculated to investigate whether revenue is reasonable relative to customer order volume.

The distribution showed:

- Mean revenue per order: approximately **₹2,490.64**
- Median revenue per order: approximately **₹1,643.46**
- Maximum revenue per order: approximately **₹57,007.40**

The large difference between the mean and median, together with the very high maximum, indicates a highly right-skewed distribution and the presence of potential high-value or anomalous observations.

The highest revenue-per-order customer generated approximately ₹57,007 per order, which is substantially higher than the typical customer. This should be investigated further during the outlier-analysis phase rather than immediately classified as bad data.

### Overall Assessment

The dataset is **usable for further analysis, but should not yet be treated as fully clean**.

The primary data-quality issues identified are:

1. **20 duplicate customer records**
2. **50 missing `discount_pct` values**
3. **Potential extreme revenue-per-order observations**

The basic business-rule validation did not identify invalid orders, returns, discounts, revenue, or tenure values.

The duplicate customer records should be handled before customer-level analysis because they can artificially inflate customer counts and revenue totals. Missing discount values should be retained or handled according to the analytical question rather than arbitrarily imputed.

The extreme revenue-per-order observations should be investigated separately to determine whether they represent legitimate high-value customers or potential data-quality issues.

**Conclusion:** The dataset is sufficiently reliable to continue the investigation after documenting the duplicate-record issue, while the missing values and potential outliers should be explicitly accounted for in subsequent analysis.
